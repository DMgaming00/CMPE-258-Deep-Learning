# Image Augmentation\n## Overview\nA/B test on image models with and without augmentation.\n

```python
import tensorflow as tf\nfrom tensorflow.keras import layers\n
```

## Image Augmentation\nUsing tf.image and Keras preprocessing layers.\n

```python
data_augmentation = tf.keras.Sequential([\n  layers.RandomFlip('horizontal'),\n  layers.RandomRotation(0.2),\n  layers.RandomZoom(0.2)\n])\n
```

## Training with Augmentation\n

```python
# Model with augmentation\nmodel = tf.keras.Sequential([data_augmentation, layers.Flatten(), layers.Dense(10)])\n
```
