## Time Series Augmentation\nNoise injection and time warping on sine wave.\n

```python
t = np.linspace(0, 50, 500)\nseries = np.sin(t) + np.random.normal(0, 0.1, 500)\n\nnoise = np.random.normal(0, 0.2, 500)\naugmented_series = series + noise\n\nplt.plot(series, label='Original')\nplt.plot(augmented_series, label='Augmented', alpha=0.7)\nplt.legend()\nplt.show()\n
```
