## PyTorch Regularization\nBaseline vs L2 & Dropout.\n

```python
class Baseline(nn.Module):\n    def __init__(self):\n        super().__init__()\n        self.seq = nn.Sequential(nn.Flatten(), nn.Linear(784, 300), nn.ReLU(), nn.Linear(300, 10))\n    def forward(self, x): return self.seq(x)\n
```

```python
class RegModel(nn.Module):\n    def __init__(self):\n        super().__init__()\n        self.seq = nn.Sequential(nn.Flatten(), nn.Dropout(0.2), nn.Linear(784, 300), nn.ReLU(), nn.Dropout(0.4), nn.Linear(300, 10))\n    def forward(self, x): return self.seq(x)\n
```
